```python
import pandas as pd
print("="*90)
print("Phase 1: Dataset Inspection \n")
print("="*90)
df1 = pd.read_csv('patient.csv')
df2 = pd.read_csv('laab_reports.csv')
df3 = pd.read_csv('oxygen_reports.csv')
md =pd.merge(df1,df2,on= "PatientID")
print(md)
print("="*90)
fd = pd.merge(md,df3,on="PatientID")
print(fd)
print("="*90)
print(fd.head(5))
print("="*90)
print(fd.tail(5))
print("="*90)
print(fd.shape)
print("="*90)
print(fd.columns)
print("="*90)
print(fd.info())
print("="*90)
print(fd.describe())

print("="*90)
print("Phase 2: Data Cleaning\n")
print("="*90)
print(fd.isnull())
print("="*90)
print(fd.fillna(0))
print("="*90)
print(fd[fd.duplicated()])
print("="*90)
print(fd.drop_duplicates())
print("="*90)
fd['FullGender']=fd['Gender'].map({
  "F":"Female",
  "M":"Male"
})
print(fd)
print("="*90)
print(fd.replace({
  "F":"Female",
  "M":"Male"
}))

print("="*90)
print("Phase 3: Feature Engineering \n")
print("="*90)

fd['AgeGroup']=fd['Age'].apply(lambda x: "senior" if (x>=60) else "Adult")
print(fd)
fd['BP_Status']=fd['BloodPressure'].apply(lambda x:"High" if (x>140) else "Normal")
print(fd)
fd['Cholesterol_Status']=fd['Cholesterol'].apply(lambda x:"High" if (x>240) else "Normal")
print(fd)

print("="*90)
print("Phase 4 : iloc Practice\n")
print("="*90)

print(fd.iloc[5,:])
print(fd.tail(5))
print(fd.iloc[:,4])
print(fd.iloc[:,[4,5]])

print("="*90)
print("Phase 5 : GroupBy Analysis\n")
print("="*90)

print(fd.groupby('Risk')['BloodPressure'].mean())
print(fd.groupby('Gender')['Cholesterol'].mean())
print(fd.groupby('Risk')['HeartRate'].max())
print(fd.groupby('Risk')['PatientID'].count())

print(fd.groupby('Risk')[['BloodPressure','HeartRate','Cholesterol']].mean())

fd['avg_BP'] = fd['BloodPressure'].mean()
print(fd)
print(fd.groupby('Risk')['avg_BP'].max())
fd['avg_Cholesterol']=fd['Cholesterol'].mean()
print(fd)
print(fd.groupby('Gender')['avg_Cholesterol'].max())
print(fd['BloodSugar'].max())

fd['high'] =fd['Risk']=="High"
print(fd['high'].count())
print(fd[fd['OxygenLevel']<95])
print(fd)

```
