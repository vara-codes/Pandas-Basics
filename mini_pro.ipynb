```python
import pandas as pd

df = pd.read_csv('patients.csv')
print(df)
print("="*50)
print(df[['Patient','HeartRate']])
print("="*50)
old = df[df['Age'] > 60]
print(old)
print("="*50)
print(df[df['Risk']=="High"])
print("="*50)
print(df[(df['Age'] > 55) & (df['BloodPressure'] > 130)])
print("="*50)
print(df.sort_values('Age',ascending=False))
print("="*50)
print(df.sort_values('HeartRate'))
print("="*50)
print("BLOOD PRESSURE ANALYSIS")
print("="*50)
print("\n1 .maximum blood pressure is: ",df['BloodPressure'].max())
print("\n2 . minimum blood pressure is: ",df['BloodPressure'].min())
print("\n3 . average blood pressure is: ",df['BloodPressure'].mean())
print("="*50)
df["HealthScore"] = df["HeartRate"]+df["BloodPressure"]
print(df)
print("="*50)
print(df.rename(columns={"HeartRate":"HR"},inplace=True))
print("="*50)
print(df.shape)
print("="*50)
print(df.info())
print("="*50)
print("CHALLENGE ACCEPTED")
print(df[(df['Risk']=="High") & (df["HR"]>80)])
print("="*50)
updated = df[['Patient','Age','Risk']].sort_values('Age')
print(updated)

print("="*50)
print("\n",df.describe())
```
