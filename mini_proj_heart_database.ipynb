```python
import pandas as pd
import numpy as np

patients = { "Patient": ["Aarav", "Diya", "Kabir", "Meera", "Rohan", "Anaya", "Aarav"],
    "Age": [65, 52, 71, 48, 60, 55, 65],
    "HeartRate": [78, 92, np.nan, 76, 88, np.nan, 78],
    "BloodPressure": [120, 145, 150, 118, 132, 140, 120],
    "Risk": ["Low", "High", "High", "Low", "Medium", "High", "Low"] }

df = pd.DataFrame(patients)
print("="*50)
print(" Cardio AI Patient Database ")
print("="*50)

print(df)
print("*"*40)
print("missing values in the dataset :", df.isnull().sum())
print("*"*40)
df["HeartRate"] = df["HeartRate"].fillna(df["HeartRate"].mean())
print(df)
print("*"*40)
print("Duplicate patient records in the dataset :", df[df.duplicated()])
print("*"*40)
print("Removed duplicate patient records in the dataset :", df.drop_duplicates())
print("*"*40)
df["Riskscore"] = df["HeartRate"] + df["BloodPressure"] 
print("updated dataset with risk score : \n", df)
print("*"*40)
print("sorted by blood pressure (descending) : \n", df.sort_values(by=["BloodPressure"], ascending=False))
print("*"*40)
risk_counts= df[df["Risk"]== "High"]
print("pat40nts with high risk : \n", risk_counts)
print("*"*40)
print("count of patients with high risk : \n",df["Risk"].value_counts())
print("*"*40)
df.rename(columns={"BloodPressure": "BP"}, inplace=True)
print(df)
print("*"*40)
print(df[(df["BP"] > 130) & (df["Age"] > 55)])

```
